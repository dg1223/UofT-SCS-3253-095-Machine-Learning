Q2 [OPTIONAL]. Explain step by step the process to select k in the k-nearest neighbor algorithm (pseudocode) 

<pre>
select a random k
(rule of thumb: k < sqrt(n), where n = number of training samples)

# Training
for each new data point:
    calculate the Euclidean distance between this sample and all training samples
    rank (sort) neighbouring samples by smallest to largest distance
    choose top-k neighbouring samples
    label new sample based on a majority vote (classification) or by taking an average (regression)

# Choosing k
1. Visualize decision boundaries:
    if the boundary snugly fits the samples -> overfitting  -> choose higher k
    if the boundary is too smooth or simple -> underfitting -> choose lower k
2. Cross-validation
    split training data into training and validation sets
    perform k-fold cross-validation on different values of k (for example, k = 1...10)
    choose the k with the lowerst validation error (percentage of mislabelled samples)
</pre>

Q4. Define a function that takes a 1-d numpy array, a parameter k, and a number p.
The function returns an estimate equal to the mean of the closest k points to the number p

```python
def k_neighbor(input_data, k, p):
    """Returns the k-neighbor estimate for p using data input_data.
    
    Keyword arguments:
    input_data -- numpy array of all the data
    k -- Number of k
    p -- input values
    
    If you make assumptions please explain them in the comments. i.e. tie breaking strategy.
    """
    
    #YOUR CODE HERE
    euclidean_distance = []
    
    # create a minheap of the euclidean distances
    for num in input_data:
        q.heappush(euclidean_distance, np.abs((p - num)))   # in 1D, Euclidean distance = |x-y|
    
    # return average of top-k values (k smallest values) from the minheap
    return np.mean(euclidean_distance[:k])
```

```python
#Evaluate
data = np.array([1,3,4,5,7,8,11,12,13,15,19,24,25,29,40])
print(k_neighbor(input_data=data, k=3, p=5))
print(k_neighbor(input_data=data, k=2, p=15))
print(k_neighbor(input_data=data, k=3, p=25))
print(k_neighbor(input_data=data, k=1, p=55))
print(k_neighbor(input_data=data, k=3, p=55))
print(k_neighbor(input_data=data, k=10, p=55))
print(k_neighbor(input_data=data, k=15, p=55)) # additional entry
print(k_neighbor(input_data=data, k=10, p=5))  # additional entry
print(k_neighbor(input_data=data, k=15, p=5))  # additional entry
```
