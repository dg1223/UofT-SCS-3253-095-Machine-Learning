Q2 [OPTIONAL]. Explain step by step the process to select k in the k-nearest neighbor algorithm (pseudocode) 

<pre>
select a random k
(rule of thumb: k < sqrt(n), where n = number of training samples)

# Training
for each new data point:
    calculate the Euclidean distance between this sample and all training samples
    rank (sort) neighbouring samples by smallest to largest distance
    choose top-k neighbouring samples
    label new sample based on a majority vote (classification) or by taking an average (regression)

# Choosing k
1. Visualize decision boundaries:
    if the boundary snugly fits the samples -> overfitting  -> choose higher k
    if the boundary is too smooth or simple -> underfitting -> choose lower k
2. Cross-validation
    split training data into training and validation sets
    perform k-fold cross-validation on different values of k (for example, k = 1...10)
    choose the k with the lowerst validation error (percentage of mislabelled samples)
</pre>

Q4. Define a function that takes a 1-d numpy array, a parameter k, and a number p.
The function returns an estimate equal to the mean of the closest k points to the number p

```python
def k_neighbor(input_data, k, p):
    """Returns the k-neighbor estimate for p using data input_data.
    
    Keyword arguments:
    input_data -- numpy array of all the data
    k -- Number of k
    p -- input values
    
    If you make assumptions please explain them in the comments. i.e. tie breaking strategy.
    """
    
    #YOUR CODE HERE
    euclidean_distance = []
    
    # create a minheap of the euclidean distances
    for num in input_data:
        q.heappush(euclidean_distance, np.abs((p - num)))   # in 1D, Euclidean distance = |x-y|
    
    # return average of top-k values (k smallest values) from the minheap
    return np.mean(euclidean_distance[:k])
```

Q5. Similar to Q4 but for the n dimentional case. 

```python
def l1_norm(a,b):
    """Returns the l1 norm (a,b)"""
    if np.shape(a) != np.shape(b):
        raise ValueError("Both arrays must have the same shape.")
    
    arr_a = np.asarray(a)
    arr_b = np.asarray(b)
    
    # calcualte manhattan distance
    return np.sum(np.abs(arr_a - arr_b))
    
def l2_norm(a,b):
    """Returns the l2 norm (a,b)"""
    if np.shape(a) != np.shape(b):
        raise ValueError("Both arrays must have the same shape.")
    
    arr_a = np.asarray(a)
    arr_b = np.asarray(b)
    
    # calculate euclidean distance
    return np.sqrt(np.sum(arr_a - arr_b)**2)
    
    
def k_neighbor_nd(input_data, k, p, metric='l1', mode='mean'):
    """Returns the k-neighbor estimate for p using data input_data.

    Keyword arguments:
    input_data -- numpy array of all the data
    k -- Number of k
    p -- input values
    metric -- l1 or l2. l1 norm or l2 norm https://en.wikipedia.org/wiki/Norm_(mathematics)
    mode -- estimator possible values = 'mean', 'median', 'max'
    
    Implement the l1 and l2 norms
    for mean, median and max, use np.mean, np.median and np.max
    """
    
    #YOUR CODE HERE
    p = np.asarray(p)
    
    distance = []
    
    # create minheap of distances (sorted from smallest to largest)
    for point in input_data:
        if metric.strip().lower() == 'l1':
            q.heappush(distance, l1_norm(p, point))
        elif metric.strip().lower() == 'l2':
            q.heappush(distance, l2_norm(p, point))
        else:
            raise ValueError('Incorrect input. Please enter either l1 or l2')    
    
    #return answer
    
    # take mean/median/max of the top k values (k smallest values)
        if mode.strip().lower() == 'mean':
            return np.mean(distance[:k])
        elif mode.strip().lower() == 'median':
            return np.median(distance[:k])
        elif mode.strip().lower() == 'max':
            return np.max(distance[:k])
        else:
            raise ValueError('Incorrect input. Please enter mean, median or max')
```

<pre>
There are two major differences between my approach and Scikit-learn's approach.

1. Scikit-learn indexes the training data using KD Tree (fewer than 20 features) or Ball Tree (higher dimensional structures) algorithms.
This is a preprocessing step that helps it calculate distances and choosing k-nearest neighbours quickly. This technique makes the function
scalable for huge training datasets. My solution does not have this step.
My solution is a brute-force approach that calculates the distance of the input to every training sample. Scikit-learn also has this mode.

2. To choose the k-nearest neighbours, I have used minheap which has a O(Nlogk) worst-case time complexity. Scikit-learn uses a combination of
techniques that gets them O(N) time complexity for this operation. They first partition the already indexed training data using a highly
vectorized operation (np.argpartition). Then they isolates the portion of the data that will contain the k-nearest neighbours, and a few more.
The data is still unsorted but it's a very small subset of the training dataset. Then they use np.argsort to sort the data in the chosen
partition to fetch k nearest neighbours.  
</pre>
